```python
from collections import deque
import time
from functools import wraps

def measure_time(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        ini = time.time()
        res = func(*args, **kwargs)
        fin = time.time()
        print(f"Tiempo de cómputo = {fin-ini} segundos")
        return res

    return wrapper

@measure_time
def pushN(pila, n):
    for i in range(n):
        pila.append(i)

pila = deque()
pushN(pila, 10000)
```

```python
# Implementar una pila con: push, pop, isempty, peek, size

class Pila:

  def __init__(self):
    self.items = deque()

  def __iter__(self):
    return reversed(self.items)

  def push(self, item):
    self.items.append(item)

  def is_empty(self):
    return self.size() == 0

  def pop(self):
    if not self.is_empty():
      self.items.pop()
    else:
      raise "index out of range"

  def peek(self):
    if not self.is_empty():
      return self.items[-1]
    else:
      raise "index out of range"

  def size(self):
    return len(self.items)
```

```python
# Implementar una cola con: push, pop, isempty, peek, size

class Cola:

  def __init__(self):
    self.items = deque()

  def __iter__(self):
    return iter(self.items)

  def push(self, item):
    self.items.append(item)

  def is_empty(self):
    return self.size() == 0

  def pop(self):
    if not self.is_empty():
      self.items.popleft()
    else:
      raise "index out of range"

  def peek(self):
    if not self.is_empty():
      return self.items[0]
    else:
      raise "index out of range"

  def size(self):
    return len(self.items)
```

```python
# Meter 1_000_000 de elementos en lista, conjunto, diccionario, pila y cola y medir tiempos

import random

@measure_time
def meter_1000_en_pila():
  pila = Pila()
  for i in range(1_000_000):
    pila.push(random.randint(0, 1_000_000))

print("Pila: ", end="")
meter_1000_en_pila()

@measure_time
def meter_1000_en_cola():
  cola = Cola()
  for i in range(1_000_000):
    cola.push(random.randint(0, 1_000_000))

print("Cola: ", end="")
meter_1000_en_cola()

@measure_time
def meter_1000_en_lista():
  lista = []
  for i in range(1_000_000):
    lista.append(random.randint(0, 1_000_000))

print("Lista: ", end="")
meter_1000_en_lista()

@measure_time
def meter_1000_en_conjunto():
  conjunto = set()
  for i in range(1_000_000):
    conjunto.add(random.randint(0, 1_000_000))

print("Conjunto: ", end="")
meter_1000_en_conjunto()

@measure_time
def meter_1000_en_diccionario():
  diccionario = {}
  for i in range(1_000_000):
    diccionario[i] = random.randint(0, 1_000_000)

print("Diccionario: ", end="")
meter_1000_en_diccionario()
```
