```python
# 1. Mapa de conexiones: {origen: {vecino: coste}}
grafo = {
    'A': {'B': 4, 'C': 2},
    'B': {'A': 4, 'C': 1, 'D': 5},
    'C': {'A': 2, 'B': 1, 'D': 8, 'E': 10},
    'D': {'B': 5, 'C': 8, 'E': 2},
    'E': {'C': 10, 'D': 2}
}

origen = 'A'
destino = 'E'

# 2. Inicialización
distancias = {nodo: float('inf') for nodo in grafo}
procedencia = {nodo: None for nodo in grafo}
pendientes = list(grafo.keys())
distancias[origen] = 0

print("+" + "-" * 70 + "+")
print(f"| RUTA OPTIMA: {origen} -> {destino}".ljust(71) + "|")
print("+" + "-" * 70 + "+")

# 3. Bucle principal
while pendientes:
    # Seleccionar el nodo no visitado con menor coste acumulado
    actual = pendientes[0]
    for nodo in pendientes:
        if distancias[nodo] < distancias[actual]:
            actual = nodo

    if distancias[actual] == float('inf') or actual == destino:
        break

    pendientes.remove(actual)

    print(f"\n[ NODO FIJADO: {actual} | Coste base acumulado: {distancias[actual]} ]")
    print("  +--------+-----------------------+------------+-----------------------+")
    print("  | Vecino | Calculo (Base + Tramo)| Estado     | Accion / Decision     |")
    print("  +--------+-----------------------+------------+-----------------------+")

    for vecino, coste in grafo[actual].items():
        nuevo = distancias[actual] + coste
        previo = distancias[vecino]
        calc_str = f"{distancias[actual]} + {coste} = {nuevo}"

        if nuevo < previo:
            distancias[vecino] = nuevo
            procedencia[vecino] = actual
            ant_str = "inf" if previo == float('inf') else str(previo)
            detalle = f"Elegido (via {actual}: {nuevo})"
            print(f"  | {vecino:<6} | {calc_str:<21} | MEJORA     | {detalle:<21} |")
        else:
            detalle = f"Descartado (queda {previo})"
            print(f"  | {vecino:<6} | {calc_str:<21} | DESCARTADO | {detalle:<21} |")

    print("  +--------+-----------------------+------------+-----------------------+")

# 4. Reconstrucción del camino
camino = []
nodo = destino
while nodo is not None:
    camino.insert(0, nodo)
    nodo = procedencia[nodo]

# 5. Resumen final
print("\n" + "+" + "-" * 70 + "+")
print("| RESUMEN FINAL".ljust(71) + "|")
print(f"|   Camino final:  {' -> '.join(camino)}".ljust(71) + "|")
print(f"|   Coste total:   {distancias[destino]}".ljust(71) + "|")
print("+" + "-" * 70 + "+")
```
