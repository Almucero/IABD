## Ejercicio 1

Crea una lista de los cuadrados de los primeros 10 números naturales.

Salida esperada: `[0, 1, 4, 9, 16, 25, 36, 49, 64, 81]`

```python
lista_cuadrados = [numero ** 2 for numero in range(10)]
print(lista_cuadrados)
```
