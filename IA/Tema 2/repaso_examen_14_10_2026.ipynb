### Ejercicio 5
A partir de las siguientes listas de desayunos y el diccionario de calorías por alimento, calcula el total de calorías de cada desayuno utilizando la función `reduce` y funciones `lambda`.

```python

desayuno2 = ['café con leche', 'donut', 'naranja']
```

### Ejercicio 5
Dada la siguiente lista de pedidos con su día de entrega y peso en kilos:
```python
pedidos = [
  ["Lunes", 3],
  ["MIÉRCOLES", 12],
  ["jueves", 25],
  ["Sábado", 8],
  ["domingo", 30],
  ["festivo", 5]
]
```
Recorre la lista y normaliza el día a minúsculas. Usa `match-case` para determinar la tarifa base según el día: lunes o martes (5€), miércoles o jueves (7€), viernes, sábado o domingo (10€), y cualquier otro día como `"Día no válido"`. Si el día es válido, usa `if-elif-else` para sumar un recargo por peso: menos de 5kg (+0€), menos de 20kg (+4€), 20kg o más (+10€).

**Salida esperada:**
```text
Pedido lunes (3kg): 5€
Pedido miércoles (12kg): 11€
Pedido jueves (25kg): 17€
Pedido sábado (8kg): 14€
Pedido domingo (30kg): 20€
Pedido festivo (5kg): Día no válido
```

````python
pedidos = [
  ["Lunes", 3],
  ["MIÉRCOLES", 12],
  ["jueves", 25],
  ["Sábado", 8],
  ["domingo", 30],
  ["festivo", 5]
]

for dia, peso in pedidos:
  dia_norm = dia.lower()

  match dia_norm:
    case "lunes" | "martes":
      tarifa_base = 5
    case "miércoles" | "miercoles" | "jueves":
      tarifa_base = 7
    case "viernes" | "sábado" | "sabado" | "domingo":
      tarifa_base = 10
    case _:
      tarifa_base = None

  if tarifa_base != None:
    if peso < 5:
      tarifa_base += 0
    elif peso < 20:
      tarifa_base += 4
    else:
      tarifa_base += 10
    print(f"Pedido {dia_norm} ({peso}kg): {tarifa_base}$")
  else:
    print(f"Pedido {dia_norm} ({peso}kg): Día no válido")

````

Invócala pasando parámetros por posición y por nombre.

```python
print(evaluar_alumno("Carlos Ruiz", [4, 6, 8, 5, 3]))
print(evaluar_alumno(notas=[3, 4, 2, 5], nombre="Lucía Gómez"))
```

**Salida esperada:**

```text
Alumno: Lucía Gómez | Media: 3.50 | Estado: Suspenso
```

````python
def evaluar_alumno(nombre: str, notas: list) -> str:
  media = sum(notas) / len(notas)
  estado = "Aprobado" if media >= 5 else "Suspenso"
  return "Alumno: {} | Media: {:.2f} | Estado: {}".format(nombre, media, estado)

print(evaluar_alumno("Carlos Ruiz", [4, 6, 8, 5, 3]))
print(evaluar_alumno(notas=[3, 4, 2, 5], nombre="Lucía Gómez"))
````

### Ejercicio 9
Crea una función `piramide_hueca(altura)` que dibuje una pirámide simétrica de asteriscos, pero donde el interior de las filas intermedias esté relleno de guiones `-` en lugar de asteriscos (la cima y la base deben ser completamente de asteriscos).

```python
piramide_hueca(5)
```

**Salida esperada:**
```text
    *
   *-*
  *---*
 *-----*
*********
```

````python
def piramide_hueca(altura: int) -> None:
  for i in range(altura):
    if i == 0:
      print(" " * (altura - 1) + "*")
    elif i == altura - 1:
      print("*" * (2 * altura - 1))
    else:
      espacios = " " * (altura - 1 - i)
      guiones = "-" * (2 * i - 1)
      print(f"{espacios}*{guiones}*")

piramide_hueca(5)
````

### Ejercicio 10
Crea una función `reloj_arena(altura_mitad)` que reciba la altura de la mitad de la figura y dibuje un reloj de arena combinando una pirámide invertida en la parte superior y una pirámide normal en la parte inferior, compartiendo el asterisco central.

```python
reloj_arena(4)
```

**Salida esperada:**
```text
*******
 *****
  ***
   *
  ***
 *****
*******
```

````python
def reloj_arena(altura_mitad: int) -> None:
  for i in range(altura_mitad):
    espacios = " " * i
    asteriscos = "*" * ((altura_mitad - i) * 2 - 1)
    print(f"{espacios}{asteriscos}")
  for i in range(altura_mitad - 2, -1, -1):
    espacios = " " * i
    asteriscos = "*" * ((altura_mitad - i) * 2 - 1)
    print(f"{espacios}{asteriscos}")

reloj_arena(4)
````

```python
puertos = [80, 443, 3306, 8080]
```
Utilizando una *list comprehension* anidada (doble `for`), genera todas las combinaciones posibles con el formato `"servidor:puerto"`, pero filtrando para que:
1. No se incluya el servidor `"srv-backup"`.
2. Solo se combinen puertos pares que no terminen en 0.

````python
servidores = ["srv-web", "srv-db", "srv-backup"]
puertos = [80, 443, 3306, 8080]

combinaciones = [
    f"{servidor}:{puerto}"
    for servidor in servidores
    for puerto in puertos
    if servidor != "srv-backup" and puerto % 2 == 0 and not str(puerto).endswith("0")
]

print(combinaciones)
````

### Ejercicio 1
Queremos simular combates entre gladiadores utilizando enumeraciones y valores aleatorios (`random.seed(42)` para que la salida sea exacta). Crea un `Enum` llamado `Arma` (`ESPADA = "espada"`, `TRIDENTE = "tridente"`) y la clase `Gladiador` con valores por defecto (`arma=Arma.ESPADA`, `energia=30`, `fuerza=80`).

El método `entrena(tipo)` incrementa la fuerza en 15 si `tipo` (sin importar mayúsculas) es `"intenso"`, o resta 5 de energía si es cualquier otro. El método `combate_con(rival)` requiere que ambos tengan al menos 15 de energía; si pelean, se calcula un desgaste aleatorio (`random.randint(5, 12)`) que se resta a la energía de ambos y un bono (`random.randint(2, 8)`) que se suma a su fuerza, ganando quien tenga más fuerza tras el bono.

Tenemos el siguiente código:
```python
import random
from enum import Enum

random.seed(42)

espartaco = Gladiador("Espartaco", arma=Arma.TRIDENTE, energia=40, fuerza=90)
crixo = Gladiador("Crixo", energia=10)

print(espartaco)
print(crixo)
espartaco.entrena("Intenso")
espartaco.combate_con(crixo)
crixo.energia = 25
espartaco.combate_con(crixo)
```

```text
Gladiador: Crixo | Arma: una espada | Energía: 19 | Fuerza: 82
```

````python
import random
from enum import Enum

random.seed(42)

class Arma(Enum):
  ESPADA = "espada"
  TRIDENTE = "tridente"

class Gladiador:
  def __init__(self, nombre: str, arma: str = Arma.ESPADA, energia: int = 30, fuerza: int = 80) -> None:
    self.nombre = nombre
    self.arma = arma
    self.energia = energia
    self.fuerza = fuerza

  def entrena(self, tipo: str) -> None:
    if tipo.lower() == "intenso":
      self.fuerza += 15
      print(f"{self.nombre} entrena duro y su fuerza sube a {self.fuerza}")
    else:
      self.fuerza -= 5
      print(f"{self.nombre} no entrana correctamente y su fuerza baja a {self.fuerza}")

  def combate_con(self, rival: "Gladiador") -> None:
    if self.energia < 15:
        print(f"{self.nombre} no tiene energía suficiente para combatir.")
        return
    if rival.energia < 15:
        print(f"{rival.nombre} no tiene energía suficiente para combatir.")
        return

    desgaste = random.randint(5, 12)
    bono = random.randint(2, 8)

    self.energia -= desgaste
    rival.energia -= desgaste
    self.fuerza += bono
    rival.fuerza += bono

    ganador, perdedor = (self.nombre, rival.nombre) if self.fuerza >= rival.fuerza else (rival.nombre, self.nombre)
    print(f"¡{ganador} vence a {perdedor}! (Desgaste: -{desgaste} energía, Bono: +{bono} fuerza)")

  def __str__(self) -> str:
    armax = "un tridente" if self.arma == Arma.TRIDENTE else "una espada"
    return(
        f"{self.__class__.__name__}: {self.nombre} | "
        f"Arma: {armax} | "
        f"Energía: {self.energia} | "
        f"Fuerza: {self.fuerza}"
    )

espartaco = Gladiador("Espartaco", arma=Arma.TRIDENTE, energia=40, fuerza=90)
crixo = Gladiador("Crixo", energia=10)

print(espartaco)
print(crixo)
espartaco.entrena("Intenso")
espartaco.combate_con(crixo)
crixo.energia = 25
espartaco.combate_con(crixo)
print(espartaco)
print(crixo)
````

Crea las subclases `NodoWeb` (con atributo `conexiones_max=100` por defecto) y `NodoBackup` (con atributo `compresion=True` por defecto, que al transferir divide los GB entre 2 usando división entera antes de sumarlos al nodo y al total global). Reutiliza el código de la superclase con `super()`.

Tenemos el siguiente código:
```python

  Servidor("192.168.1.1"),
  NodoWeb("192.168.1.10", conexiones_max=500),
  NodoBackup("192.168.1.20", compresion=True)
]

for equipo in flota_red:
  print(f"\n[{equipo.__class__.__name__}]")
  equipo.transfiere(120)
  print(equipo)

```

**Salida esperada:**
```text

[NodoWeb]

IP: 192.168.1.10 | Tráfico: 120 GB | Conexiones máx: 500

Comprimiendo bloques de datos...
IP: 192.168.1.20 | Tráfico: 60 GB | Compresión: Sí
```

````python
class Servidor:
  trafico_total: int = 0

  def __init__(self, ip: str) -> None:
    self.ip = ip
    self.trafico: int = 0

  def transfiere(self, gb: int) -> None:
    if gb < 0:
      raise ValueError("El tráfico no puede ser negativo.")
    self.trafico += gb
    Servidor.trafico_total += gb

  def __str__(self) -> str:
    return f"IP: {self.ip} | Tráfico: {self.trafico} GB"

  @classmethod
  def muestra_trafico_total(cls) -> None:
    print(f"Tráfico total del centro de datos: {cls.trafico_total} GB.")

class NodoWeb(Servidor):
  def __init__(self, ip: str, conexiones_max: int = 100) -> None:
    super().__init__(ip)
    self.conexiones_max = conexiones_max

  def transfiere(self, gb: int) -> None:
    print("Procesando peticiones HTTP...")
    super().transfiere(gb)

  def __str__(self) -> str:
    return f"{super().__str__()} | Conexiones máx: {self.conexiones_max}"

class NodoBackup(Servidor):
  def __init__(self, ip: str, compresion: bool = True) -> None:
    super().__init__(ip)
    self.compresion = compresion

  def transfiere(self, gb: int) -> None:
    print("Comprimiendo bloques de datos...")
    gb_transferir = gb // 2 if self.compresion else gb
    super().transfiere(gb_transferir)

  def __str__(self) -> str:
    compresion_texto = "Sí" if self.compresion else "No"
    return f"{super().__str__()} | Compresión: {compresion_texto}"

flota_red = [
  Servidor("192.168.1.1"),
  NodoWeb("192.168.1.10", conexiones_max=500),
  NodoBackup("192.168.1.20", compresion=True)
]

for equipo in flota_red:
  print(f"\n[{equipo.__class__.__name__}]")
  equipo.transfiere(120)
  print(equipo)

Servidor.muestra_trafico_total()
````

Una pasarela criptográfica gestiona monederos digitales protegidos mediante encapsulación. La clase `Monedero` recibe el `propietario` y los `fondos_iniciales` (por defecto 0), almacenándolos en un atributo privado `__fondos` accesible solo para lectura mediante `@property` (`fondos`), además de llevar un registro de clase del volumen total depositado en la plataforma.

* Si `fondos_iniciales < 0`, o si las cantidades en `deposita(cantidad)` o `transfiere_a(destino, cantidad)` son `<= 0`, lanza `ValueError("Cantidad no válida.")`.
* Si en `transfiere_a` no hay fondos suficientes, lanza `ValueError("Fondos insuficientes.")`.

Tenemos el siguiente código:
```python
m_alvaro = Monedero("Álvaro", 500)
m_elena = Monedero("Elena")

m_alvaro.deposita(150)
m_alvaro.transfiere_a(m_elena, 200)

print(f"Fondos de {m_alvaro.propietario}: {m_alvaro.fondos} €")
print(f"Fondos de {m_elena.propietario}: {m_elena.fondos} €")

try:
  m_elena.transfiere_a(m_alvaro, 300)
except ValueError as e:
  print(f"Error en transferencia: {e}")

Monedero.muestra_volumen_depositos()
```

**Salida esperada:**
```text
Fondos de Álvaro: 450 €
Fondos de Elena: 200 €
Error en transferencia: Fondos insuficientes.
Volumen total depositado en la plataforma: 650 €
```

````python
class Monedero:
  volumen_total_depositos: float = 0

  def __init__(self, propietario: str, fondos_iniciales: float = 0) -> None:
    if fondos_iniciales < 0:
      raise ValueError("Cantidad no válida.")
    self.propietario = propietario
    self.__fondos = fondos_iniciales
    Monedero.volumen_total_depositos += fondos_iniciales

  @property
  def fondos(self) -> float:
    return self.__fondos

  def deposita(self, cantidad: float) -> None:
    if cantidad <= 0:
      raise ValueError("Cantidad no válida.")
    self.__fondos += cantidad
    Monedero.volumen_total_depositos += cantidad

  def transfiere_a(self, destino: "Monedero", cantidad: float) -> None:
    if cantidad <= 0:
      raise ValueError("Cantidad no válida.")
    if cantidad > self.__fondos:
      raise ValueError("Fondos insuficientes.")
    self.__fondos -= cantidad
    destino.__fondos += cantidad

  @classmethod
  def muestra_volumen_depositos(cls) -> None:
    print(f"Volumen total depositado en la plataforma: {cls.volumen_total_depositos} €")

m_alvaro = Monedero("Álvaro", 500)
m_elena = Monedero("Elena")

m_alvaro.deposita(150)
m_alvaro.transfiere_a(m_elena, 200)

print(f"Fondos de {m_alvaro.propietario}: {m_alvaro.fondos} €")
print(f"Fondos de {m_elena.propietario}: {m_elena.fondos} €")

try:
  m_elena.transfiere_a(m_alvaro, 300)
except ValueError as e:
  print(f"Error en transferencia: {e}")

Monedero.muestra_volumen_depositos()
````

Cada vez que una mascota pasa por `consulta(variacion_peso)`:
* Si el peso resultante (`__peso + variacion_peso`) es `<= 0`, lanza `ValueError("Peso resultante inválido.")`.
* Los perros pagan 30 € base más 2 € por cada kilo de su peso **actualizado** tras la variación.
* Las aves pagan una tarifa plana de 20 €, o 35 € si son exóticas (`exotica=True`).

Tenemos el siguiente código:
```python

  Perro("Rex", peso_inicial=15.0, raza="Pastor Alemán"),
  Ave("Paco", peso_inicial=0.5, exotica=True),
  Perro("Chispa", peso_inicial=4.0)
```

````python
class Mascota:
  _recaudacion_total: float = 0.0

  def __init__(self, nombre: str, peso_inicial: float) -> None:
    self.nombre = nombre
    self.__peso = float(peso_inicial)

  @property
  def peso(self) -> float:
    return self.__peso

  def consulta(self, variacion_peso: float) -> None:
    nuevo_peso = round(self.__peso + variacion_peso, 2)
    if nuevo_peso <= 0:
      raise ValueError("Peso resultante inválido.")
    self.__peso = nuevo_peso

  @classmethod
  def muestra_recaudacion(cls) -> float:
    return cls._recaudacion_total

class Perro(Mascota):
  def __init__(self, nombre: str, peso_inicial: float, raza: str = "Mestizo") -> None:
    super().__init__(nombre, peso_inicial)
    self.raza = raza

  def consulta(self, variacion_peso: float) -> None:
    super().consulta(variacion_peso)
    coste = 30.0 + 2.0 * self.peso
    Mascota._recaudacion_total += coste

  def __str__(self) -> str:
    return f"[{self.__class__.__name__}] Nombre: {self.nombre} | Peso: {self.peso:.1f} Kg | Raza: {self.raza}"

class Ave(Mascota):
  def __init__(self, nombre: str, peso_inicial: float, exotica: bool = False) -> None:
    super().__init__(nombre, peso_inicial)
    self.exotica = exotica

  def consulta(self, variacion_peso: float) -> None:
    super().consulta(variacion_peso)
    coste = 35.0 if self.exotica else 20.0
    Mascota._recaudacion_total += coste

  def __str__(self) -> str:
    exotica_str = "Sí" if self.exotica else "No"
    return f"[{self.__class__.__name__}] Nombre: {self.nombre} | Peso: {self.peso:.1f} Kg | Exótica: {exotica_str}"

pacientes = [
  Perro("Rex", peso_inicial=15.0, raza="Pastor Alemán"),
  Ave("Paco", peso_inicial=0.5, exotica=True),
  Perro("Chispa", peso_inicial=4.0)
]

pacientes[0].consulta(1.0)
pacientes[1].consulta(0.1)
pacientes[2].consulta(-0.5)

for p in pacientes:
  print(p)

print(f"Recaudación total de la clínica: {Mascota.muestra_recaudacion()} €")
````

### Ejercicio 6
Dada la siguiente lista de lecturas de temperatura en grados Celsius:
```python
temperaturas = [-5, 12, -8, 0, 22, -15, 7, -3, 18, -10]
```
Resuelve el problema en tres pasos encadenados usando programación funcional:
1. Usa `filter` para quedarte solo con las temperaturas bajo cero (`< 0`) que sean impares.
2. Usa `map` para transformar cada temperatura filtrada en su cuadrado (`** 2`).
3. Usa `reduce` para obtener el producto (multiplicación acumulada, con valor inicial `1`) de todos los cuadrados resultantes.

**Salida esperada:**
```text
Filtradas: [-5, -15, -3]
Cuadrados: [25, 225, 9]
Producto total: 50625
```

````python
temperaturas = [-5, 12, -8, 0, 22, -15, 7, -3, 18, -10]

temperaturas_filtradas = list(filter(lambda temperatura: temperatura < 0 and temperatura % 2 != 0, temperaturas))
print(f"Filtradas: {temperaturas_filtradas}")

temperaturas_mapeadas = list(map(lambda temperatura: temperatura ** 2, temperaturas_filtradas))
print(f"Cuadrados: {temperaturas_mapeadas}")

producto_cuadrados = reduce(lambda acumulado, elemento: acumulado * elemento, temperaturas_mapeadas, 1)
print(f"Producto total: {producto_cuadrados}")
````

### Ejercicio 7
Una tienda online gestiona el siguiente catálogo de artículos:

```python
catalogo = [

  {"codigo": "B05-STD", "precio": 50, "unidades": 0, "iva": 10},
  {"codigo": "C99-PRO", "precio": 200, "unidades": 2, "iva": 21},
  {"codigo": "D40-LITE", "precio": 80, "unidades": 5, "iva": 4},
  {"codigo": "E10-PRO", "precio": 150, "unidades": 1, "iva": 10}
]
```
Utiliza `filter`, `map` y `reduce` para calcular la facturación total (con IVA incluido) de los artículos que tengan unidades disponibles (`unidades > 0`) y cuyo código termine en `"-PRO"`.
*(Fórmula del importe por artículo: `precio * unidades * (1 + iva / 100)`).*

**Salida esperada:**
```text
Artículos válidos: [{'codigo': 'A12-PRO', 'precio': 100, 'unidades': 3, 'iva': 21}, {'codigo': 'C99-PRO', 'precio': 200, 'unidades': 2, 'iva': 21}, {'codigo': 'E10-PRO', 'precio': 150, 'unidades': 1, 'iva': 10}]
Importes con IVA: [363.0, 484.0, 165.0]
Facturación total: 1012.0
```

````python
catalogo = [
  {"codigo": "A12-PRO", "precio": 100, "unidades": 3, "iva": 21},
  {"codigo": "B05-STD", "precio": 50, "unidades": 0, "iva": 10},
  {"codigo": "C99-PRO", "precio": 200, "unidades": 2, "iva": 21},
  {"codigo": "D40-LITE", "precio": 80, "unidades": 5, "iva": 4},
  {"codigo": "E10-PRO", "precio": 150, "unidades": 1, "iva": 10}
]

catalofo_filtrado = list(filter(lambda articulo: articulo["unidades"] > 0 and articulo["codigo"].endswith("-PRO"), catalogo))
print(f"Artículos válidos: {catalofo_filtrado}")

importes_iva = list(map(lambda articulo: articulo["precio"] * articulo["unidades"] * (1 + articulo["iva"] / 100), catalofo_filtrado))
print(f"Importes con IVA: {importes_iva}")

facturacion_total = reduce(lambda acumulado, importe: acumulado + importe, importes_iva, 0)
print(f"Facturación total: {facturacion_total}")
````
