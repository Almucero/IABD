# Programación orientada a objetos en Python

Python es un lenguaje de prpósito general y multiparadigma. Permite, entre otros enfoques, la programación imperativa, la progrmación funcional y la **programación orientada a objetos (POO)**

## Programación orientada a objetos (POO)

* Es un paradigma de programación que se basa en la utilización de objetos.
* Se caracteriza por dar importancia a los datos (más que al flujo del programa).
* Los datos del mundo real se modelan mediante clases.
* Un programa consiste en "la vida y obra" de los objetos creados a partir de esas clases: acciones de esos objetos sobre sí mismos y sobre otros objetos.

## Clase

Es un modelo que se crea a partir de una simplificación de un concepto del mundo real a partir del cual se crean objetos. Define que características van a tener esos objetos y que acciones tienen asociadas.

## Objeto o instancia

Deriva de la concrección de una clase.

Una clase sería "alumno", mientras que "Ana Del Bosque" y "Pepito Pérez" serían objetos de esa clase.

Los datos asociados a cada objeto se almacenan en **atributos de instancia**. Por ejemplo, cada coche puede tener su propia matrícula, marca, modelo o color.

Las acciones asociadas a los objetos se definen mediante **métodos de instancia**. Por ejemplo, un coche podría arrancar, detenerse o reocorrer una distancia.

## `self` y `__init__()`

En los métodos de instancia, `self` representa el objeto sobre el que se está trabajando.

El método especial `__init_()` (inicializador) se ejecuta inmediatamente después de crear un objeto y se utiliza habitualmente para inicializar sus atributos.

El constructor real es `__new__()` pero en Python se usa en casos muy puntuales

```python
class GatoBasico:
  """Representa un gato de forma muy simplificada"""

  def __init__(self, nombre, color="marrón", edad=1):
    self.nombre = nombre
    self.color = color
    self.edad = edad

  def maulla(self):
    print("Miauuuuuuu")

  def ronronrea(self):
    print("Mrrrrrrrrr")
```

```python
garfield = GatoBasico("Garfield", color="naranja", edad=6)
lisa = GatoBasico("Lisa Marisa")

print(garfield.nombre, garfield.color, garfield.edad)
print(lisa.nombre, lisa.color, lisa.edad)
```

```python
print(garfield)
```

## El método `__str__()`

Este método especial permite definir una representación textual legible de un objeto. Es el equivalente al método `toString()` de Java.

```python
from enum import Enum
import random

class Sexo(Enum):
  MACHO = "macho"
  HEMBRA = "hembra"

class Gato:
  """Representa un gato y algunas de sus funciones"""

  def __init__(
      self,
      nombre: str,
      color: str = "marrón",
      raza: str = "mestizo",
      sexo: Sexo = Sexo.HEMBRA,
      edad: int = 1,
      peso: float = 1.5,
      energia: int = 50,
      fuerza: int = 100,
  ) -> None:
    self.nombre = nombre
    self.color = color
    self.raza = raza
    self.sexo = sexo
    self.edad = edad
    self.peso = peso
    self.energia = energia
    self.fuerza = fuerza

  def __str__(self) -> str:
    gatx = "una gata" if self.sexo == Sexo.HEMBRA else "un gato"
    return (
        f"Hola, soy {self.nombre}, {gatx} de color {self.color}, "
        f"de raza {self.raza}. Tengo {self.edad} años y "
        f"peso {self.peso} Kg."
    )

  def maulla(self) -> None:
    print("Miauuuuuuu")

  def ronronrea(self) -> None:
    print("Mrrrrrrrrr")

  def come(self, comida: str) -> None:
    """Acepta pescado y rechaza cualquier otra comida"""
    print(f"Peso de {garfield.nombre} antes de comer: {garfield.peso}")

    if (comida.lower() == "pescado"):
      print(f"{self.nombre} se come el pescado y engorda.")
      self.peso = self.peso + 1
    else:
      print(f"{self.nombre} no come {comida} y pierde peso.")
      self.peso = self.peso - 1

    print(f"Peso de {garfield.nombre} después de comer: {garfield.peso}")

  def saluda_a(self, otro: "Gato") -> None:
    """Hace que un gato salude a otro"""
    print(f"{self.nombre} frota la cabeza contra {otro.nombre}")

  def pelea_con(self, contrincante: "Gato") -> None:
      """Simula una pelea entre dos gatos.

      Solo hay pelea si ambos tienen al menos 10 puntos de energía.
      El gato con más fuerza gana y, tras la pelea, ambos pierden
      energía y ganan algo de fuerza.
      """
      print(f"Energía de {self.nombre} antes del combate: {self.energia}")
      print(f"Fuerza de {self.nombre} antes del combate: {self.fuerza}")
      print(f"Energía de {contrincante.nombre} antes del combate: {contrincante.energia}")
      print(f"Fuerza de {contrincante.nombre} antes del combate: {contrincante.fuerza}\n")

      perdida_energia = random.randint(1, 5)
      ganancia_fuerza = random.randint(1, 10)

      if self.energia >= 10 and contrincante.energia >= 10:
        self.energia -= perdida_energia
        contrincante.energia -= perdida_energia
        self.fuerza += ganancia_fuerza
        contrincante.fuerza += ganancia_fuerza

        if self.fuerza > contrincante.fuerza:
          print(f"{self.nombre} gana el combate contra {contrincante.nombre}")
        elif contrincante.fuerza > self.fuerza:
          print(f"{contrincante.nombre} gana el combate contra {self.nombre}")
        else:
          print("La pelea ha acabado en un empate. ")

        print(
            f"Ambos han perdido {perdida_energia} puntos de energía "
            f"y han ganado {ganancia_fuerza} puntos de fuerza\n"
        )
      else:
        if self.energia < 10 and contrincante.energia < 10:
          print("Ninguno de los dos gatos tiene energía para pelear")
        elif self.energia < 10:
          print(f"{self.nombre} no tiene energía para pelear.")
        else:
          print(f"{contrincante.nombre} no tiene energía para pelear.")

      print(f"Energía de {self.nombre} después del combate: {self.energia}")
      print(f"Fuerza de {self.nombre} después del combate: {self.fuerza}")
      print(f"Energía de {contrincante.nombre} después del combate: {contrincante.energia}")
      print(f"Fuerza de {contrincante.nombre} después del combate: {contrincante.fuerza}")
```

```python
garfield = Gato("Garfield", color="naranja", edad=6, sexo=Sexo.MACHO, fuerza=600, energia=55)
lisa = Gato("Lisa Marisa", fuerza=100)
```

```python
print(garfield)
print(lisa)
```

```python
garfield.come("Pescado")
```

```python
garfield.saluda_a(lisa)
```

```python
garfield.pelea_con(lisa)
```

## Atributos y métodos de clase

Los atributos de instancia pertenecen a cada objeto. Por ejemplo, cada vehículo tendrá su propio kilometraje.

Un **atributo de clase**, sin embargo, pertenece a la clase y es compartido por sus instancias. Es útil para almacenar información común, como el kilometraje acumulado de todos los vehículos.

Un **método de clase** se aplica a una clase, no a un objeto.

Un método decorado con `@classmethod` recibe la clase como primer argumento, que por convenio es `cls`.

```python
class Vehiculo:
  """Representa un vehículo y su kilometraje."""

  kilometraje_total: int = 0

  def __init__(self, marca: str, modelo: str) -> None:
    self.marca: str = marca
    self.modelo: str = modelo
    self.kilometraje: int = 0

  def __str__(self) -> None:
    return(
        f"Marca: {self.marca} | Modelo: {self.modelo} | "
        f"Kilometraje: {self.kilometraje} Km"
    )

  def recorre(self, km: int) -> None:
    """Incrementa el kilometraje individual y acumulado."""
    if km < 0:
      raise ValueError("La distancia no puede ser negatviva.")

    self.kilometraje += km
    Vehiculo.kilometraje_total += km

  @classmethod
  def muestra_kilometraje_total(cls) -> None:
    print(
        f"Entre todos los vehículos han recorrido"
        f" {cls.kilometraje_total} km."
    )
```

```python
vehiculo1 = Vehiculo("Cupra", "Raval")
vehiculo2 = Vehiculo("Seat", "León")
vehiculo3 = Vehiculo("Honda", "PCX")

vehiculo1.recorre(500)
vehiculo2.recorre(200)
vehiculo3.recorre(75)
vehiculo2.recorre(100)

print(vehiculo1)
print(vehiculo2)
print(vehiculo3)

Vehiculo.muestra_kilometraje_total()
```

## Herencia

La herencia permite crear una clase nueva a partir de otra existente. La nueva clase, llamada **subaclase**, hereda los atributos y métodos de su superclase y puede añadir comportamiento o modificarlo.

En Python, `super()` permite invocar la implementación definida en la superclase.

```python
class Coche(Vehiculo):
  """La clase Coche se define como subclase de Vehiculo"""

  def __init__(self, marca: str, modelo: str, numero_de_puertas: int = 4) -> None:
    super().__init__(marca, modelo)
    self.numero_de_puertas = numero_de_puertas

  def __str__(self) -> str:
    return (
      f"{super().__str__()} | "
      f"Nº de puertas: {self.numero_de_puertas}"
    )

  def recorre(self, km: int) -> None:
    super().recorre(km)
    print("Estoy quemando gasolina")

class Bicicleta(Vehiculo):
  """La clase Bicicleta es subclase de Vehiculo"""

  def __init__(self, marca: str, modelo: str, numero_de_pinones: int = 7) -> None:
    super().__init__(marca, modelo)
    self.numero_de_pinones = numero_de_pinones

  def __str__(self) -> str:
    return (
      f"{super().__str__()} | "
      f"Nº de piñones: {self.numero_de_pinones}"
    )

  def recorre(self, km: int) -> None:
    super().recorre(km)
    print("Avanzo gracias al pedaleo")
```

```python
coche = Coche("Toyota", "Celica", numero_de_puertas=3)
bicicleta = Bicicleta("Orbea", "Furia", 1)

coche.recorre(50)
bicicleta.recorre(10)

print(coche)
print(bicicleta)

Vehiculo.muestra_kilometraje_total()
```

## Polimorfismo

Permite trabajar con objetos de clases diferentes usando el mismo nombre de método.

```python
flota = [
  Vehiculo("Kawasaki", "GPZ 500"),
  Coche("Dacia", "Duster", numero_de_puertas=5),
  Bicicleta("Orbea", "Alma", numero_de_pinones=12),
]

for vehiculo in flota:
  print(f"\n{vehiculo.__class__.__name__}")
  vehiculo.recorre(5)
  print(vehiculo)
```

## Encapsulación en Python

Python no impone niveles de acceso como `privado`, `protected` y `public` al estilo de Java. En su lugar utiliza **convecniones**.

* `atributo`: forma pública normal.
* `_atributo`: el equivalente a protegido.
* `__atributo`: el equivalente a privado.

Cuando queremos controlar el acceso a un dato, la solución es utilizar `@property`.

```python
class CuentaBancaria:
  """Ejemplo sencillo de encapsulación"""

  def __init__(self, titular: str, saldo_inicial: float = 0) -> None:
    if saldo_inicial < 0:
      raise ValueError("El saldo inicial no puede ser negativo")

    self.titular = titular
    self.__saldo = saldo_inicial

  @property
  def saldo(self) -> float:
    """Devuelve el saldo actual sin permitir asignarlo directamente"""
    return self.__saldo

  def ingresa(self, cantidad: float) -> None:
    if cantidad <= 0:
      raise ValueError("La cantidad a ingresar debe ser positiva.")
    self.__saldo += cantidad

  def retira(self, cantidad: float) -> None:
    if cantidad <= 0:
      raise ValueError("La cantidad a retirar debe ser positiva.")
    if cantidad > self.__saldo:
      raise ValueError("Saldo insuficiente.")
    self.__saldo -= cantidad
```

```python
cuenta = CuentaBancaria("Ana Pérez", saldo_inicial=100)
cuenta.ingresa(50)
cuenta.retira(30)
print(cuenta.saldo)
```

La propiedad `saldo` permite consultar el dato como si fuera un atributo normal (`cuenta.saldo`), pero su modificación queda controlada por los métodos `ingresa()` y `retira()`.
