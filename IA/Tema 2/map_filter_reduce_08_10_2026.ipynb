# Map, Filer y Reduce

Las funciones `map()`, `filter()` y `reduce()` son extremadamente potentes y, bien utilizadas, ayudan a reducir las líneas de un programa y a dejar el código más limpio.

Están basadas en el paradigma de la programación funcional.

Se aplican recorriendo estructuras iterables (por ej. listas) y realizando determinadas acciones sobre esos iterables.

## Funciones de orden superior (higher-order functions)

Una función de orden superior es aquella que recibe otra función como parámetro o devuelve una función como resultado.

````python
def decora_con_estrellas(texto):
  return f"⭐⭐⭐{texto}⭐⭐⭐"

def triplica(texto):
  return (texto + " ") * 3

def no_hace_nada(texto):
  return texto

def saluda(funcion, nombre):
  print(funcion(f"Hola {nombre}!"))
````

````python
saluda(decora_con_estrellas, "Álvaro Jiménez")
````

````python
saluda(triplica, "Álvaro Jiménez")
````

````python
saluda(no_hace_nada, "Álvaro Jiménez")
````

````python
def crea_multiplicado(n):
  def multiplica(x):
    return x * n

  return multiplica # Devuelve una función
````

````python
duplica = crea_multiplicado(2)
triplica = crea_multiplicado(3)
````

````python
print(duplica(100))
print(triplica(500))
````

## Funciones lambda

Son expresiones que permiten definir funciones anónimas con un formato muy conciso.

Una función lambda solo puede contener una expresión. Si necesitamos varias sentencias o una lógica más compleja, es mejor recurrir a las funciones tradicionales.

````python
def par(n):
  return (n % 2) == 0

print(par(7))
print(par(22))
````

````python
par = lambda n: (n % 2) == 0

print(par(7))
print(par(22))
````

## Map

Transforma cada uno de los elmentos del iterable.

````python
def cubo(n):
  return n ** 3

numeros = [5, 7, 22, 97, 54, 62, 77, 23, 73, 61]

numeros_al_cubo = list(map(cubo, numeros))

print(numeros)
print(numeros_al_cubo)
````

````python
numeros_al_cubo = list(map(lambda n: n ** 3, numeros))

print(numeros)
print(numeros_al_cubo)
````

En Python, `map()` devuelve un iterador, no una lista.

````python
numeros_al_cubo = map(lambda n: n ** 3, numeros)

print(numeros_al_cubo)
````

````python
print(next(numeros_al_cubo))
print(next(numeros_al_cubo))
````

````python
for numero in numeros_al_cubo:
  print(numero)
````

## Filter

Filtra los elementos de un iterable en función de si sus elementos cumple o no una condición.

````python
# Filtra los números para obtener solo los positivos (mayores que cero)

numeros = [-1, 5, -7, 22, 97, 0, 54, 62, -77, 23, 73, -61]
print(numeros)

positivos = list(filter(lambda numero: numero > 0, numeros))
print(positivos)
````

palabras = ["casa", "asa", "mesa", "PHP", "COBOL", "ADA", "ser", "somos"]

## Reduce

La función `reduce` sirve para consumir los elementos de un iterable mediante una función y obtener un valor.

Se obtiene como resultado un único elemento que puede ser de cualquier tipo.

La función que se utiliza para consumir elementos tiene dos parámetros, el primero es el acumulado de lo que se lleva consumiendo y el segundo es el elemento que se va sacando.

El formato es el siguiente:

```
reduce(funcion_de_consumo, iterable, valor_inicial)
```

````python
from functools import reduce

numeros = [1, 5, 7, 22, 97, 0, 54, 62, 77, 23, 73, 61]

suma = reduce(lambda acumulado, elemento: acumulado + elemento, numeros, 0)
print(suma)
````

## Ejercicio

Utiliza `reduce` para sumar todos los números impares de una lista.

````python
numeros = [1, 5, 7, 22, 97, 0, 54, 62, 77, 23, 73, 61]

suma_impares = reduce(lambda acumulado, elemento: acumulado + (0 if elemento % 2 == 0 else elemento), numeros, 0)
print(suma_impares)
````

````python
numeros = [1, 5, 7, 22, 97, 0, 54, 62, 77, 23, 73, 61]
impares = list(filter(lambda impar: impar % 2 != 0, numeros))

suma_impares = reduce(lambda acumulado, elemento: acumulado + elemento, impares, 0)
print(suma_impares)
````

## Combinando `map`,  `filter` y `reduce`

````python
# Suma de los cuadrados de los números pares

numeros = [1, 2, 3, 4, 5, 6]
print(f"Números: {numeros}")

pares = list(filter(lambda numero: numero % 2 == 0, numeros))
print(f"Pares: {pares}")

cuadrados = list(map(lambda numero: numero ** 2, pares))
print(f"Cuadrados: {cuadrados}")

suma = reduce(lambda acumulado, elemento: acumulado + elemento, cuadrados, 0)
print(f"Suma: {suma}")
````

## Ejercicio

Dada la siguiente lista de productos, utiliza `filter`, `map` y `reduce` para calcular el valor total del stock de los productos de los que quedan unidades.

```
productos = [
    {"nombre": "Ratón", "precio": 15, "stock": 10},
    {"nombre": "Teclado", "precio": 25, "stock": 0},
    {"nombre": "Monitor", "precio": 180, "stock": 4},
    {"nombre": "USB", "precio": 8, "stock": 20}
]
```

````python
productos = [
    {"nombre": "Ratón", "precio": 15, "stock": 10},
    {"nombre": "Teclado", "precio": 25, "stock": 0},
    {"nombre": "Monitor", "precio": 180, "stock": 4},
    {"nombre": "USB", "precio": 8, "stock": 20}
]

productos_filtrados = list(filter(lambda producto: producto["stock"] > 0, productos))
print(f"Productos filtrados: {productos_filtrados}")

productos_mapeados = list(map(lambda producto: producto["precio"] * producto["stock"], productos_filtrados))
print(f"Productos mapeados: {productos_mapeados}")

valor_total = reduce(lambda acumulado, elemento: acumulado + elemento, productos_mapeados, 0)
print(f"Valor total: {valor_total}")
````
