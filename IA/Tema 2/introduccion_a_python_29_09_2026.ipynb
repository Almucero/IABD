## Sentencia condicional

```python
x = 88

if x == 88:
  print("La variable x vale 88")
  print("Es un número bonito")
else:
  print("La variable x tiene un valor distinto a 88")
```

```python
edad = int(input("Dime tu edad: "))

if edad < 18:
  print("¡A estudiar!")
elif edad < 24:
  print("Sería bueno que aprendieras una profesión")
elif edad < 66:
  print("¡A trabajar!")
else:
  print("¿Te gusta mirar las obreas del barrio?")
```

```python
dia = input("Dime el día de la semana (por ej. lunes): ").lower()

cadena = "Ese día, a cuarta hora toca "

match dia:
  case "lunes" | "martes":
    print(cadena + "Big Data.")
  case "miércoles" | "sábado" | "domingo":
    print("¡Día libre!")
  case "jueves":
    print(cadena + "Inteligencia Artificial.")
  case "viernes":
    print(cadena + "Libre, por poco.")
  case _:
    print("Ese día no existe.")
```

Es un trozo de código encapsulado que realiza una tarea específica y que se puede invocar desde cualquier sitio de la aplicación.

```python
def saluda():
  print("Hola")
```

```python
saluda()
```

```python
def saluda(nombre):
  print("Hola, {}.".format(nombre))
```

```python
def media(x, y, z):
  resultado = (x + y + z) / 3
  return resultado
```

```python
x = media(12, 15, 33)
print(x)
```

```python
def media(numeros):
  resultado = sum(numeros) / len(numeros)
  return resultado
```

```python
mi_lista = [4, 8, 10, 22, 33, 55]
print(media(mi_lista))
```

```python
print(media([4, 8, 10, 22]))
```
