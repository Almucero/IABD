## Variables

Los nombres de variables se escriben en formato *snake case*.

Por ejemplo, sería correcto escribir precio_sin_iva. No sería correcto precioSinIva.



```python
x = 88
print(x)
print(type(x))
```

```python
pi = 3.141592654
print(pi)
print(type(pi))
```

```python
fruta = "mandarina"
frutero = fruta * 7
```

```python
print("x vale:", x)
print ("pi vale:", pi)
print("Mi fruta preferida es:", fruta)
print("En la mesa hay", frutero)
```

## Operadores

```python
print(x + pi)
print(x * pi)
print(x ** 2)
dos_frutas = fruta + fruta
frutero = fruta * 5
print(dos_frutas)
print(frutero)
```

```python
print(7 / 2) # división con decimales
print(7 // 2) # división entera
print(7 % 2) # resto de la división (módulo)
```

```python
print(f"La variable x multiplicada por pi vale {x * pi:.2f}.")
```
