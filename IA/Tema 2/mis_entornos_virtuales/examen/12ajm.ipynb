### Ejercicio 3

Usa `map` para clasificar los  siguientes tiempos de respuesta de una aplicación:

```python
tiempos = [45, 130, 210, 90, 150]
```

según estos criterios:

- Menos de 100 ms: `"rápida"`.
- Entre 100 y 200 ms, ambos incluidos: `"normal"`.
- Más de 200 ms: `"lenta"`.

**Salida esperada:**
```python
['rápida', 'normal', 'lenta', 'rápida', 'normal']
```

````python
tiempos = [45, 130, 210, 90, 150]

def criterio(elemento):
    if elemento < 100:
        return "rápida"
    elif elemento < 200:
        return "normal"
    else:
        return "lenta"

salida = list(map(lambda elemento: criterio(elemento), tiempos))

print(salida)
````

### Ejercicio 7

Usa `reduce()` para calcular la suma de todos los dígitos de la siguiente lista de matrículas:

```python
matriculas = ["1234BCF", "4567DFG", "7891GJK", "2345KLM", "6789MNP", "3456PRT", "9012STV"]
```

Debes usar un `reduce()` para calcular la suma de los dígitos de cada matrícula y otro `reduce()` para calcular la suma total. No se puede usar `sum()`.

**Salida esperada:**
```python
131
```

````python
from functools import reduce

matriculas = ["1234BCF", "4567DFG", "7891GJK", "2345KLM", "6789MNP", "3456PRT", "9012STV"]

def suma_digitos(matricula):
    return reduce(lambda acumulado, elemento: acumulado + int(elemento if elemento.isdigit() else 0), matricula, 0)

suma_total = reduce(lambda acumulado, elemento: acumulado + suma_digitos(elemento), matriculas, 0)

print(suma_total)
````

### Ejercicio 11

Una empresa calcula el coste de sus envíos. Define la clase `Paquete` con estas características:

- Cada paquete tiene un **destinatario** y un **peso** en kilogramos. El peso debe ser un atributo privado.
- Define la **propiedad** de solo lectura `peso`.
- Hay dos datos que son únicos para toda la clase y que permiten calcular el coste del envío: la **tarifa_base** (vale 4 por defecto) y la **tarifa por kg** (2 por defecto).
- Define el método de clase `actualiza_tarifas(nueva_base, nueva_tarifa_kg)` para actualizar esas dos tarifas.
- El coste del envío se calcula sumando la tarifa base con la tarifa por kilogramo multiplicada por el peso.

**Código de prueba:**

```python
paquete1 = Paquete("María", 2)
print(
    f"El paquete para {paquete1.destinatario} tiene un peso de {paquete1.peso} kg"
    f"y su coste de envío es de {paquete1.coste_envio()} euros."
    )
Paquete.actualiza_tarifas(6, 3)
paquete2 = Paquete("Alejandro", 5)
print(
    f"El paquete para {paquete2.destinatario} tiene un peso de {paquete2.peso} kg"
    f"y su coste de envío es de {paquete2.coste_envio()} euros."
    )
```

**Salida esperada:**

```
El paquete para María tiene un peso de 2 kg y su coste de envío es de 8 euros.
El paquete para Alejandro tiene un peso de 5 kg y su coste de envío es de 21.75 euros.
```

````python
class Paquete:
    tarifa_base = 4
    tarifa_por_kg = 2

    def __init__(self, destinatario, peso_inicial):
        self.destinatario = destinatario
        self.__peso = peso_inicial

    @classmethod
    def actualiza_tarifas(cls, nueva_base, nueva_tarifa_kg):
        cls.tarifa_base = nueva_base
        cls.tarifa_por_kg = nueva_tarifa_kg

    @property
    def peso(self):
        return self.__peso

    def coste_envio(self):
        return f"{self.tarifa_base + self.tarifa_por_kg * self.__peso}"

paquete1 = Paquete("María", 2)
print(
    f"El paquete para {paquete1.destinatario} tiene un peso de {paquete1.peso} kg "
    f"y su coste de envío es de {paquete1.coste_envio()} euros."
    )
Paquete.actualiza_tarifas(6, 3)
paquete2 = Paquete("Alejandro", 5)
print(
    f"El paquete para {paquete2.destinatario} tiene un peso de {paquete2.peso} kg "
    f"y su coste de envío es de {paquete2.coste_envio()} euros."
    )
````

### Ejercicio 12

Crea `PaqueteUrgente` como subclase de `Paquete`:

- Cada paquete urgente tiene un **recargo**, que por defecto son 5 euros.
- Ahora, para calcular el coste del envío hay que sumar el recargo.
- Usa `super()` siempre que sea posible.

**Código de prueba:**

```python
Paquete.actualiza_tarifas(5, 2.5)
envios = [
    Paquete("Lucía", 2),
    PaqueteUrgente("Pedro", 1),
    PaqueteUrgente("Mar", 3, recargo=7)
]
for envio in envios:
    print(
    f"El paquete para {envio.destinatario} tiene un peso de {envio.peso} kg"
    f" y su coste de envío es de {envio.coste_envio()} euros."
    )
```

**Salida esperada:**
```
El paquete para Lucía tiene un peso de 2 kg y su coste de envío es de 10.0 euros.
El paquete para Pedro tiene un peso de 1 kg y su coste de envío es de 12.5 euros.
El paquete para Mar tiene un peso de 3 kg y su coste de envío es de 19.5 euros.
```

````python
class PaqueteUrgente(Paquete):
    def __init__(self, destinatario, peso_inicial, recargo=5):
        super().__init__(destinatario, peso_inicial)
        self.recargo = recargo

    def coste_envio(self):
        return float(super().coste_envio()) + float(self.recargo)

Paquete.actualiza_tarifas(5, 2.5)
envios = [
    Paquete("Lucía", 2),
    PaqueteUrgente("Pedro", 1),
    PaqueteUrgente("Mar", 3, recargo=7)
]
for envio in envios:
    print(
    f"El paquete para {envio.destinatario} tiene un peso de {envio.peso} kg"
    f" y su coste de envío es de {envio.coste_envio()} euros."
    )
````
