# Análisis de sentimientos

Usaremos un pipeline de transformers:

https://huggingface.co/docs/transformers/main_classes/pipelines

```python
from transformers import pipeline
```

```python
# Crea el pipeline
pipe = pipeline("sentiment-analysis", model="pysentimiento/robertuito-sentiment-analysis")
```

```python
# Texto que queremos analizar
# text = "My students are amazing!"
text = "¡Mis estudiantes son maravillosos!"
```

```python
# Aplica el pipeline para ver si el texto es positivo o negativo
result = pipe(text)[0]
result
```

```python
# Muestra el resultado
print(f"El texto \"{text}\" se ha clasificado como {result["label"]} con una puntuación de {round(result["score"], 4) * 100}%")
```
