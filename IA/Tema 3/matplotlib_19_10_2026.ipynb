## Importación

```python
import matplotlib.pyplot as plt
```

## Pyplot vs OO

```python
# Pyplot: traza y muestra una línea simple
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y) # plot indica el tipo de gráfica, en este caso una línea
plt.show()
```

```python
# OO: misma gráfica con objetos figure/axes
fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

## Títulos y etiquetas

```python
plt.plot(x, y)
plt.title("Cuadrados")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
# OO
fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_title("Cuadrados")
ax.set_xlabel("x")
ax.set_ylabel("y")
plt.show()
```

```python
import numpy as np

t = np.linspace(0, 2 * np.pi, 50) # array de 50 números entre 0 y 2π
s = np.sin(t) # seno de t

# la x sería la t en radianes, y la y sería el seno
plt.plot(t, s)
plt.title("Seno")
plt.xlabel("t (rad)")
plt.ylabel("seno(t)")
plt.show()
```

## Personalización de las líneas

```python
plt.plot(x, y, color="green", marker="x", linestyle=":")
plt.title("Gráfica personalizada")
plt.xlabel("Eje x")
plt.ylabel("Eje y")
plt.show()
```

```python
plt.plot(x, y, color="red", marker="o", linestyle="--")
plt.title("Gráfica personalizada")
plt.xlabel("Eje x")
plt.ylabel("Eje y")
plt.show()
```

## Gráficas superpuestas

```python
x = [1, 2, 3, 4, 5]
cuadrados = [1, 4, 9, 16, 25]
cubos = [1, 8, 27, 64, 125]

plt.plot(x, cuadrados, label="y = x²")
plt.plot(x, cubos, label="y = x³")
plt.legend()
plt.show()
```

```python
# Ejercicio
# Gráficas del seno y el coseno juntas

x = np.linspace(0, 4 * np.pi, 120)
seno = np.sin(x)
coseno = np.cos(x)

plt.plot(x, seno, label="y = seno(x)")
plt.plot(x, coseno, label="y = cos(x)")
plt.legend()
plt.show()
```

## Límites y ticks

```python
plt.plot(t, s)
plt.xlim(0, 2 * np.pi)
plt.ylim(-1.2, 1.2)
plt.show()
```

## Cuadrícula (grid)

```python
plt.plot(t, s)
plt.xlim(0, 2 * np.pi)
plt.ylim(-1.2, 1.2)
plt.xticks([0, np.pi, 2 * np.pi], ["0", "π", "2π"]) # Posiciones y etiquetas
plt.minorticks_on()
plt.grid(which="both", linestyle="--", color="lightgray")
plt.show()
```

## Gráfica de dispersión o nube de puntos (scatter)

```python
x = np.random.uniform(low=-10, high=10, size=50) # Genero 50 números aleatorios entre -10 y 10
y = np.random.uniform(low=-10, high=10, size=50)

plt.scatter(x, y)
plt.title("Puntos aleatorios entre -10 y 10")
plt.show()
```

```python
plt.title("Puntos aleatorios entre -10 y 10")
hipotenusa = np.hypot(x, y) # Calcula la hipotenusa (distancia al origen)
nube_de_puntos = plt.scatter(x, y, c=hipotenusa)
plt.colorbar(nube_de_puntos)
plt.show()
```

## Histogramas (hist)

```python
# Genera 500 números aleatorios siguiendo una distribución normal con media 0 y desviación típica 1
datos = np.random.normal(size=500)
plt.hist(datos, bins=20, edgecolor="black") # 20 barras
plt.show()
```

```python
# Histograma con edades de la población

# Genera 1000 edades con media 35 y desviación típica 10
datos = np.random.normal(loc=35, scale=10, size=1000)
plt.hist(datos, bins=12, color="green", edgecolor="black")
plt.show()
```

```python
# Histograma 2D
edades = np.random.normal(40, 10, 1000) # 1000 casos con media 40 y desviación típica 10
salarios = np.random.normal(25000, 6000, 1000) # Media 2500 €, desviación típica 6000 €
plt.hexbin(edades, salarios, gridsize=30, cmap="Blues")
plt.colorbar(label="Número de personas")
plt.xlabel("Edad")
plt.ylabel("Salario")
plt.title("Distribución de edades y salarios")
plt.show()
```

## Varias gráficas en la msima figura

```python
plt.figure(figsize=(7, 5)) # Crea una figura (lienzo) de 7 x 5 pulgadas

# Primera gráfica (fila 1, columna 1)
plt.subplot(2, 2, 1)
plt.plot(np.sin(x))
plt.title("Seno")

# Segunda gráfica (fila 1, columna 2)
plt.subplot(2, 2, 2)
plt.plot(np.cos(x))
plt.title("Coseno")

# Tercera gráfica (fila 2, columna 1)
plt.subplot(2, 2, 3)
plt.plot(np.tan(x))
plt.title("Tangente")

# Cuarta gráfica (fila 2, columna 2)
plt.subplot(2, 2, 4)
plt.plot(np.sin(x) * np.cos(x))
plt.title("Seno • Coseno")

plt.tight_layout() # Ajusta los espacios
plt.show()
```

## Ejes compartidos

```python
fig, axs = plt.subplots(2, 1, figsize=(6, 4))
axs[0].plot(np.sin(x)); axs[0].set_title("Compartiendo el eje x")
axs[1].plot(np.cos(x)); axs[1].set_xlabel("x")
plt.tight_layout() # Ajusta los espacios
plt.show()
```

```python
# Pyplot
dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"]
ventas_zapatillas = [10, 15, 8, 12, 14]
ventas_camisetas = [20, 22, 19, 24, 21]

plt.fill_between(dias, ventas_camisetas, color="salmon", alpha=0.4, label="Camisetas")
plt.fill_between(dias, ventas_zapatillas, color="skyblue", alpha=0.8, label="Zapatillas")
plt.legend(loc="upper left")
plt.title("Ventas durante la semana")
plt.show()
```

## Filtros sobre imágenes

```python
import matplotlib.pyplot as plt
from skimage.color import rgb2gray

# Cargar la imagen
gato = plt.imread("/content/drive/MyDrive/gato.jpg")

# Convertir a escala de grises
gris = rgb2gray(gato)

# Mapas de colores
mapas = [
    None, "inferno", "plasma", "viridis", "cividis", "twilight", "twilight_shifted",
    "turbo", "berlin", "managua", "vanimo", "Blues", "BrBG", "BuGn", "BuPu", "magma"
]

# Crear la figura
plt.figure(figsize=(16, 16))
plt.suptitle("Gato multicolor")

for i, mapa in enumerate(mapas):
  plt.subplot(4, 4, i + 1)
  plt.imshow(gris, cmap=mapa)
  plt.title(str(mapa))
  plt.axis("off")

plt.tight_layout()
plt.savefig("gato_multicolor.png", dpi=300, bbox_inches="tight")
plt.show()
```
