## Ejercicio 8

Muestra en una misma gráfica, las funciones lineal, cuadrática y cúbica.

Debe aparecer una leyenda indicando a qué corresponde cada línea.

La gráfica debe tener un título.

Las líneas deben ser cada una de un tipo: continua, con pequeñas líneas discontinuas y con puntos.

```python
x = np.linspace(0, 5, 100)

plt.plot(x, x, '-', label='lineal')
plt.plot(x, x**2, '--', label='cuadrática')
plt.plot(x, x**3, ':', label='cúbica')

plt.title("Gráfica de funciones")
plt.legend()
plt.show()
```

## Ejercicio 9

Representa, en la misma gráfica, una línea a base de círculos, otra hecha con triángulos y otra formada por cuadraditos.

```python
x = np.linspace(0, 5, 10)

plt.plot(x, x, 'ro')
plt.plot(x, x+1, 'g^')
plt.plot(x, x+2, 'bs')

plt.show()
```

## Ejercicio 10

Crea una gráfica de color amarillo para la función $y = x² + 1$ en el intervalo de -1 a 1 usando la API orientada a objetos de Matplotlib.

```python
x = np.linspace(-1, 1, 100)
y = x**2 + 1

fig, ax = plt.subplots()
ax.plot(x, y, color='y')
plt.show()
```

## Ejercicio 12

En esta ocasión, sólo se simulan 10 edades. Crea un diagrama de barras a partir de esas edades.

```python
edades = np.random.randint(16, 51, 10)

plt.bar(range(len(edades)), edades)
plt.show()
```

## Ejercicio 13

En un restaurante italiano han contabilizado el tipo de comida que se pide en un día. 22 personas han pedido pizza, 16 pidieron pasta, 12 tomaron lasaña y 4 comieron risotto.

Representa esta información mediante un diagrama de tarta.

```python
cantidades = [22, 16, 12, 4]
etiquetas = ['Pizza', 'Pasta', 'Lasaña', 'Risotto']

plt.figure(figsize=(6,6))
plt.pie(cantidades, labels=etiquetas)
plt.show()
```
