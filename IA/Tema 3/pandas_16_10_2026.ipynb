## Carga de datos desde un fichero

```python
iadb_df = pd.read_csv("/content/drive/MyDrive/grupo_master_iabd.csv")
```

```python
iadb_df
```

## Borrado de columnas

```python
iadb_df.drop(columns=["dni_numero", "dni_letra"], inplace=True)
```

```python
iadb_df
```

```python
iadb_df.drop(columns=["apellidos", "nombre"], inplace=True)
```

## Exploración del DataFrame

```python
iadb_df.info()
```

```python
iadb_df.describe()
```

```python
# Listado con los nombres y las notas
iadb_df[["nombre_completo", "nota_primer_trimestre", "nota_segundo_trimestre", "nota_tercer_trimestre"]]
```

```python
campos_a_mostrar = ["nombre_completo", "nota_primer_trimestre", "nota_segundo_trimestre", "nota_tercer_trimestre"]
iadb_df[campos_a_mostrar]
```

```python
iadb_df.sort_values(by="nombre_completo")
```

```python
iadb_df.sort_values(by="nombre_completo", inplace=True)
```

```python
# Creamos la columna nota_media con la media de las notas de los tres trimestres
iadb_df["nota_media"] = (iadb_df["nota_primer_trimestre"] + iadb_df["nota_segundo_trimestre"] + iadb_df["nota_tercer_trimestre"]) / 3
```

```python
iadb_df["nota_boletin"] = round(iadb_df["nota_media"])
```

```python
iadb_df
```

```python
iadb_df["nota_boletin"] = iadb_df["nota_boletin"].astype(int)
```

```python
iadb_df.info()
```

```python
# Mostrar todos los géneros que tenemos
iadb_df["genero"].unique()
```

```python
# Cambio de géneros al español
iadb_df["genero"].replace({"Male": "Masculino", "Female": "Femenino", "Polygender": "Poligénero", "Bigender": "Bigénero", "Non-binary": "No binario"}, inplace=True)
```

```python
iadb_df
```

```python
# Listado de aprobados (nombre y nota)
iadb_df[iadb_df["nota_boletin"] >= 5][["nombre_completo", "nota_boletin"]]
```

```python
# Nombre, nota media y nota del boletín de los que han sacado notable o sobresaliente
iadb_df[iadb_df["nota_boletin"] >= 7][["nombre_completo", "nota_media", "nota_boletin"]]
```

```python
# Igual pero ordenados por nota (de mayor a menor)
iadb_df[iadb_df["nota_boletin"] >= 7][["nombre_completo", "nota_media", "nota_boletin"]].sort_values(by="nota_boletin", ascending=False)
```
